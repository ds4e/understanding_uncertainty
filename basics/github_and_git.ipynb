# GitHub and Git

## Introduction
- Git is free and open source version control software: It's like a manually operated Dropbox or Box or Google Drive, that we operate to track changes on a project and collaborate
- A directory that is tracked by Git is called a **repository** or **repo**
- GitHub is a web-based service that hosts Git repos
- These notes and video are just about the bare minimum

## Basic Workflow
1. Create/Fork on GitHub
2. `git clone repo` to create your local copy
3. Work: Add new files with `git add file.ext`, save snapshots with `git commit -am 'Commit message` `
4. `git push origin main`
5. `git pull --rebase` to update your local copy

## 1. Create/Fork on GitHub
- This happens on GitHub
- Your life is easier if you imagine that projects come into existence on GitHub through creation or forking
- The repo on GitHub is called `origin`, and is the ground truth of the project (**not** what is on your computer)

## 2. `git clone repo`

- To create a **local copy** of the project to work on, `git clone https://github.com/user/repo`
- If it doesn't work, the repo is probably not public or you are not logged in with the correct rights to access it

## 3. Work: `git add` and `git commit`

1. Add new files to be tracked with `git add file.ext` or `git add directory`, 
2. Save project snapshots with `git commit -am 'Commit message' `


## Useful Commands
- `git status` : What branch are you on? What are the notable statuses of various files?
- `git ls-files` : List the files in the repo being tracked by Git
- `git rm --cached file.ext` : Stop tracking a file, but don't delete it
- `git log` : What is the history of commits for this repo?
- `git add *` : Add all untracked files in this directory to your git repo (use with caution)

## 4. Push results to GitHub: `git push origin main`
- You need to solve your authentication puzzle somehow: VS Code has a log-in button, the `gh` package is my favorite, or you can use the GitHub desktop app
- Once you have authenticated yourself to GitHub, you can push your results back: `git push origin main`
- If you don't like the result, there's a snapshot of every version of the project on the web page, and you can extract an older one and restart from there (or `git revert`)

## 5. Update locally: `git pull --rebase`
- `git pull` updates your local copy of the project from `origin`, but if you're working with other people, mutual changes can create problems with the vanilla `git pull`
- Adding `--rebase` pulls the changes from `origin main`, applies them to your local repo, and then overlays your local changes on top if possible

## `.gitignore`
- `.gitignore` lists files that GitHub should **not** track: Large data files, virtual environments, Python runtime cruft, API keys, etc.
- Get this file sorted out **before** you ever `git add *`

## Workflow

1. Create/Fork on GitHub to create `origin` for your project
2. Create local copy/branch: `git clone repo`
3. Work: `git add file.ext` and `git commit -am 'Commit message` `
4. Push changes: `git push origin main`
5. Update locally: `git pull --rebase` to update, and return to step 3

## Staging
You can control what goes in a commit, instead of just snapshoting the whole current directory

- After a commit, the "staging area" is empty of changes
- You can deliberately add files to the staging area with `git add file.ext`
- Then, commit only the staged changes with `git commit -m 'Message' `, rather than `-am`
- If you want to stage all changes, you can `git add .`

## ... Don't Push to Main
- As soon as the basic GitHub Cycle feels natural, you want to stop pushing to `main`: `main` is the sacred, stable, true version of the project
- Instead, we create branches locally, experiment, and then push the branch back to GitHub to be merged

## Branching

In order to work on a local copy of a copy without creating project_v2_final_FINAL, you use **branches**

- `git branch <name>` creates a new branch
- `git branch` lists the existing branches and currently selected branch
- `git checkout <name>` or `git switch <name>` checks out the `<name>` branch
- `git status` lists the current branch
- `git switch -c <name>` creates the new branch <name> and checks it out immediately

## Pushing a Branch
- `git push -u origin <branch_name>`
- Notice that the changes are detected on GitHub, and you create a **pull request** to merge your changes into main
- This is the safe way to update collaborative work without changing the ground truth on everyone

## Branching Workflow

1. Create/Fork on GitHub to create `origin` for your project
2. Create local copy/branch: `git clone repo` and `git switch -c <branch>`
3. Work: `git add file.ext` and `git commit -am 'Commit message' ` to your branch
4. Push changes: `git push -u origin <branch>`
5. Update locally: `git switch main`, `git pull --rebase` to update, `git switch -c <new branch>` to start a new branch for new work, and return to step 3
