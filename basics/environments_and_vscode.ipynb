## Environments and VS Code

> ## What is the problem?
>
> - Different projects have different software dependencies
> - You might even need different versions of Python
> - This can be really frustrating, and the details can depend on your specific computer (OS, versions of python, etc.)
> - At the CLI, these details are explicit, but in an IDE, they can be hidden and lead to confusion
>

> ## Pip
> - Pip handles package management at the level of the directory/repo
> - Pip (typically) creates a hidden directory called `.venv`
> - You can select it as a kernel or activate it at the command line
>- In a notebook, select it as a kernel and then type `! pip install pandas numpy matplotlib seaborn` into code chunk and run it to install packages
> - If the venv is activated at the CLI,  `pip install <package>`
> 
> $ \ $

> ## Conda
> - Conda handles packages globally
>    1. `conda init <name>` to create a new Conda virtual environment
>    2. `conda activate <name>` to turn on the virtual environment; no `<name>` selects `(base)`
>    3. `conda install <package>`; if you have a conda virtual environment on and use `pip install`, it will install into that Conda environment
>   4. `conda deactivate` to turn off the virtual environment
>- If Conda is activated at the command line, so it says `(base)` or `(<package>)`, using pip will installl packages into that Conda environment
>
